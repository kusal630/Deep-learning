```python
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.metrics import classification_report, confusion_matrix

print("gpu devices:", tf.config.list_physical_devices('GPU'))

# task 1
# kaggle copy of cifar-10 is in the old pickle format, so the batches are loaded manually
data_dir = '/kaggle/input/datasets/pankrzysiu/cifar10-python/cifar-10-batches-py'

xs, ys = [], []
for i in range(1, 6):
    with open(os.path.join(data_dir, 'data_batch_%d' % i), 'rb') as f:
        b = pickle.load(f, encoding='bytes')
    xs.append(b[b'data'])
    ys.append(b[b'labels'])

x_train = np.concatenate(xs)
y_train = np.concatenate(ys)

with open(os.path.join(data_dir, 'test_batch'), 'rb') as f:
    tb = pickle.load(f, encoding='bytes')
x_test = tb[b'data']
y_test = np.array(tb[b'labels'])

with open(os.path.join(data_dir, 'batches.meta'), 'rb') as f:
    meta = pickle.load(f, encoding='bytes')
labels = [c.decode() for c in meta[b'label_names']]

# batches store flat vectors in channel first order, converting to channels last
x_train = x_train.reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1)
x_test = x_test.reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1)
print('train:', x_train.shape, 'test:', x_test.shape)

plt.figure(figsize=(14, 3))
for i in range(10):
    plt.subplot(1, 10, i+1)
    plt.imshow(x_train[i])
    plt.axis('off')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.bar(labels, np.bincount(y_train))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# task 2: feature map sizes for different kernels with valid padding
sample = x_train[:1] / 255.0
for k in (3, 5, 7):
    out = layers.Conv2D(16, k, padding='valid')(sample)
    print('kernel %dx%d -> %s' % (k, k, out.shape))

# task 3: effect of stride and padding on the output dimensions
for s in (1, 2):
    for p in ('same', 'valid'):
        out = layers.Conv2D(16, 3, strides=s, padding=p)(sample)
        print('stride %d, %s -> %s' % (s, p, out.shape))

# task 4: visualize what the first conv layer sees
inp = layers.Input(shape=(32, 32, 3))
c1 = layers.Conv2D(32, 3, activation='relu')(inp)
p1 = layers.MaxPooling2D(2)(c1)
feat_model = models.Model(inp, c1)

maps = feat_model.predict(sample, verbose=0)[0]
plt.figure(figsize=(12, 6))
for i in range(8):
    plt.subplot(2, 4, i+1)
    plt.imshow(maps[:, :, i], cmap='magma')
    plt.axis('off')
plt.tight_layout()
plt.show()

# task 5 and 6
x_tr = x_train / 255.0
x_te = x_test / 255.0

def build_cnn(pool='max'):
    m = models.Sequential()
    m.add(layers.Conv2D(32, 3, activation='relu', input_shape=(32, 32, 3)))
    m.add(layers.MaxPooling2D(2) if pool == 'max' else layers.AveragePooling2D(2))
    m.add(layers.Conv2D(64, 3, activation='relu'))
    m.add(layers.MaxPooling2D(2) if pool == 'max' else layers.AveragePooling2D(2))
    m.add(layers.Flatten())
    m.add(layers.Dense(64, activation='relu'))
    m.add(layers.Dense(10, activation='softmax'))
    m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return m

print('training max pooling model')
net_max = build_cnn('max')
hist_max = net_max.fit(x_tr, y_train, batch_size=32, epochs=20, validation_split=0.1, verbose=2)

print('training average pooling model')
net_avg = build_cnn('avg')
hist_avg = net_avg.fit(x_tr, y_train, batch_size=32, epochs=20, validation_split=0.1, verbose=2)

# parameter count for the results table
net_max.summary()

def plot_curves(h, name):
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(h.history['accuracy'], label='train')
    plt.plot(h.history['val_accuracy'], label='val')
    plt.title(name + ' accuracy')
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(h.history['loss'], label='train')
    plt.plot(h.history['val_loss'], label='val')
    plt.title(name + ' loss')
    plt.legend()
    plt.tight_layout()
    plt.show()

plot_curves(hist_max, 'max pooling')

plot_curves(hist_avg, 'average pooling')

# task 7
y_pred = np.argmax(net_max.predict(x_te, verbose=0), axis=1)
print(classification_report(y_test, y_pred, target_names=labels))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('predicted')
plt.ylabel('actual')
plt.tight_layout()
plt.show()

print('ex1 output size:', (64 - 5 + 2*2)/2 + 1)

# 2. (3*3*3 + 1 bias) * 64 filters = 1792
print('ex2 parameters:', (3*3*3 + 1) * 64)

```
