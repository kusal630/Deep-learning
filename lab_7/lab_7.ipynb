```python
# =============================================================
# Experiment 7: Autoencoders on MNIST  (FC-AE, CAE, Denoising CAE, VAE)
# Kaggle-ready | Single-GPU optimal
# =============================================================
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'
os.environ['TF_FORCE_GPU_ALLOW_GROWTH'] = 'true'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from skimage.metrics import structural_similarity as ssim_fn
import time, warnings
warnings.filterwarnings('ignore')

# ------------------ GPU CONFIG -------------------------------
# MNIST-scale models do NOT benefit from MirroredStrategy across 2 GPUs —
# gradient-sync overhead makes it slower. We pin to GPU:0 for max speed.
gpus = tf.config.list_physical_devices('GPU')
for g in gpus:
    try:
        tf.config.experimental.set_memory_growth(g, True)
    except Exception:
        pass
print(f"GPUs detected: {len(gpus)} | Using single GPU (/GPU:0) — optimal for this workload.")

# ------------------ SEED -------------------------------------
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ------------------ LOAD DATA --------------------------------
# ------------------ LOAD DATA (robust to header/no-header CSVs) ------------
# ------------------ LOAD DATA (bulletproof) --------------------
TRAIN_CSV = '/kaggle/input/datasets/meenalsinha/mnist-handwritten-digits-csv-format/mnist_train.csv'
TEST_CSV  = '/kaggle/input/datasets/meenalsinha/mnist-handwritten-digits-csv-format/mnist_test.csv'

def load_mnist_csv(path):
    # 1. Read raw, no header assumption
    df = pd.read_csv(path, header=None, low_memory=False)

    # 2. Force every cell to numeric. Any string ('label', 'pixel0', BOM, etc.)
    #    becomes NaN — we then drop those rows.
    df = df.apply(pd.to_numeric, errors='coerce')

    # 3. Drop any row that contained a non-numeric value (i.e. the header row)
    df = df.dropna(axis=0, how='any').reset_index(drop=True)

    # 4. Ensure exactly 785 columns (1 label + 784 pixels)
    assert df.shape[1] == 785, f"Expected 785 cols, got {df.shape[1]} in {path}"

    return df.astype(np.float32)

train_df = load_mnist_csv(TRAIN_CSV)
test_df  = load_mnist_csv(TEST_CSV)

# Sanity check
print("train_df shape:", train_df.shape, "| sample label:", int(train_df.iloc[0, 0]))
print("test_df  shape:", test_df.shape,  "| sample label:", int(test_df.iloc[0, 0]))
print("Any NaN left?", train_df.isna().any().any(), test_df.isna().any().any())

# First column = label, remaining 784 = pixels
y_train_all = train_df.iloc[:, 0].values.astype(np.int32)
x_train_all = train_df.iloc[:, 1:].values.astype(np.float32) / 255.0
y_test_all  = test_df.iloc[:, 0].values.astype(np.int32)
x_test_all  = test_df.iloc[:, 1:].values.astype(np.float32) / 255.0

# Lab subset (as required by the PDF)
N_TRAIN, N_TEST = 10000, 2000
x_train, y_train = x_train_all[:N_TRAIN], y_train_all[:N_TRAIN]
x_test,  y_test  = x_test_all[:N_TEST],  y_test_all[:N_TEST]

# Reshapes
x_train_flat = x_train.reshape(-1, 784)
x_test_flat  = x_test.reshape(-1, 784)
x_train_img  = x_train.reshape(-1, 28, 28, 1)
x_test_img   = x_test.reshape(-1, 28, 28, 1)

# Train/val split (last 1000 for validation)
n_val = 1000
x_tr_flat, x_val_flat = x_train_flat[n_val:], x_train_flat[:n_val]
x_tr_img,  x_val_img  = x_train_img[n_val:],  x_train_img[:n_val]

print(f"Train: {x_tr_flat.shape} | Val: {x_val_flat.shape} | Test: {x_test_flat.shape}")
print(f"Label range — train: [{y_train.min()},{y_train.max()}]  test: [{y_test.min()},{y_test.max()}]")
# ------------------ METRICS ----------------------------------
def compute_metrics(orig, recon):
    """orig, recon: (N,28,28) or (N,28,28,1) float [0,1]."""
    o = orig.reshape(-1, 28, 28)
    r = np.clip(recon.reshape(-1, 28, 28), 0.0, 1.0)
    mse = float(np.mean((o - r) ** 2))
    mae = float(np.mean(np.abs(o - r)))
    ssims = [ssim_fn(o[i], r[i], data_range=1.0) for i in range(len(o))]
    return mse, mae, float(np.mean(ssims))

def per_image_mse(orig, recon):
    o = orig.reshape(-1, 784)
    r = np.clip(recon.reshape(-1, 784), 0.0, 1.0)
    return np.mean((o - r) ** 2, axis=1)

def show_grid(images, titles, n=10, figsize=(15, 4), suptitle=""):
    images = list(images)
    plt.figure(figsize=figsize)
    for i in range(n):
        for j, img in enumerate(images):
            plt.subplot(len(images), n, j * n + i + 1)
            plt.imshow(img[i].squeeze(), cmap='gray', vmin=0, vmax=1)
            plt.axis('off')
            if i == 0:
                plt.title(titles[j], fontsize=9)
    if suptitle:
        plt.suptitle(suptitle, fontsize=11)
    plt.tight_layout()
    plt.show()

# =============================================================
# 1. FULLY-CONNECTED AUTOENCODER
# =============================================================
def build_fc_ae(latent_dim=16):
    inp = layers.Input(shape=(784,), name='input')
    x   = layers.Dense(128, activation='relu')(inp)
    x   = layers.Dense(32,  activation='relu')(x)
    z   = layers.Dense(latent_dim, activation='relu', name='latent')(x)
    x   = layers.Dense(32,  activation='relu')(z)
    x   = layers.Dense(128, activation='relu')(x)
    out = layers.Dense(784, activation='sigmoid')(x)
    return Model(inp, out, name=f'FC_AE_z{latent_dim}')

fc_ae = build_fc_ae(16)
fc_ae.compile(optimizer=Adam(1e-3), loss='binary_crossentropy')
t0 = time.time()
hist_fc = fc_ae.fit(
    x_tr_flat, x_tr_flat,
    validation_data=(x_val_flat, x_val_flat),
    epochs=20, batch_size=128, verbose=2
)
fc_train_time = time.time() - t0

# ---- Plot 1: Original vs Reconstructed ----
def show_grid(images, titles, n=10, figsize=(15, 4), suptitle=""):
    images = list(images)
    plt.figure(figsize=figsize)
    for i in range(n):
        for j, img in enumerate(images):
            plt.subplot(len(images), n, j * n + i + 1)
            a = np.asarray(img[i])
            # Flattened 784 -> reshape to 28x28 so imshow is happy
            if a.size == 784:
                a = a.reshape(28, 28)
            else:
                a = a.squeeze()
            plt.imshow(a, cmap='gray', vmin=0, vmax=1)
            plt.axis('off')
            if i == 0:
                plt.title(titles[j], fontsize=9)
    if suptitle:
        plt.suptitle(suptitle, fontsize=11)
    plt.tight_layout()
    plt.show()
# ---- Plot 2: Training / Val loss ----
plt.figure(figsize=(7, 4))
plt.plot(hist_fc.history['loss'], label='Train')
plt.plot(hist_fc.history['val_loss'], label='Val')
plt.xlabel('Epoch'); plt.ylabel('Binary Cross-Entropy')
plt.title('Plot 2 — FC-AE Training and Validation Loss')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

fc_mse, fc_mae, fc_ssim = compute_metrics(x_test, fc_recon)
print(f"[FC-AE] MSE={fc_mse:.5f}  MAE={fc_mae:.5f}  SSIM={fc_ssim:.4f}  "
      f"Params={fc_ae.count_params():,}  Time={fc_train_time:.1f}s")

# =============================================================
# 2. CONVOLUTIONAL AUTOENCODER
# =============================================================
def build_cae(latent_dim=None):
    inp = layers.Input(shape=(28, 28, 1))
    x = layers.Conv2D(32, 3, activation='relu', padding='same')(inp)
    x = layers.MaxPooling2D(2, padding='same')(x)              # 14x14
    x = layers.Conv2D(64, 3, activation='relu', padding='same')(x)
    x = layers.MaxPooling2D(2, padding='same')(x)              # 7x7
    x = layers.Conv2D(64, 3, activation='relu', padding='same', name='latent')(x)
    if latent_dim is not None:                                  # optional dense bottleneck
        x = layers.Flatten()(x)
        x = layers.Dense(latent_dim, activation='relu', name='latent_vec')(x)
        x = layers.Dense(7 * 7 * 64, activation='relu')(x)
        x = layers.Reshape((7, 7, 64))(x)
    x = layers.UpSampling2D(2)(x)                               # 14x14
    x = layers.Conv2D(32, 3, activation='relu', padding='same')(x)
    x = layers.UpSampling2D(2)(x)                               # 28x28
    out = layers.Conv2D(1, 3, activation='sigmoid', padding='same')(x)
    return Model(inp, out, name='CAE')

cae = build_cae()   # follows PDF's exact architecture (conv feature-map bottleneck)
cae.compile(optimizer=Adam(1e-3), loss='binary_crossentropy')
t0 = time.time()
hist_cae = cae.fit(
    x_tr_img, x_tr_img,
    validation_data=(x_val_img, x_val_img),
    epochs=20, batch_size=128, verbose=2
)
cae_train_time = time.time() - t0

cae_recon = cae.predict(x_test_img, batch_size=256, verbose=0)
cae_mse, cae_mae, cae_ssim = compute_metrics(x_test, cae_recon)
print(f"[CAE]   MSE={cae_mse:.5f}  MAE={cae_mae:.5f}  SSIM={cae_ssim:.4f}  "
      f"Params={cae.count_params():,}  Time={cae_train_time:.1f}s")

# ---- Plot 3: FC vs CAE ----
show_grid([x_test, fc_recon, cae_recon],
          ["Original", "FC-AE", "CAE"],
          n=10, figsize=(18, 3.5),
          suptitle="Plot 3 — Original vs FC-AE vs CAE Reconstruction")

# =============================================================
# 3. DENOISING CONVOLUTIONAL AUTOENCODER
# =============================================================
def add_gaussian(imgs, sigma):
    n = np.random.normal(0, sigma, imgs.shape).astype(np.float32)
    return np.clip(imgs + n, 0.0, 1.0)

def add_salt_pepper(imgs, p):
    out = imgs.copy()
    r = np.random.random(imgs.shape)
    out[r < p / 2] = 0.0
    out[(r >= p / 2) & (r < p)] = 1.0
    return out

# Build paired dataset for denoising (input=noisy, target=clean)
def make_denoise_dataset(clean_imgs, sigma=0.2, batch=128):
    noisy = add_gaussian(clean_imgs, sigma)
    ds = tf.data.Dataset.from_tensor_slices((noisy, clean_imgs))
    return ds.shuffle(2000, seed=SEED).batch(batch).prefetch(tf.data.AUTOTUNE)

denoiser = build_cae()
denoiser.compile(optimizer=Adam(1e-3), loss='binary_crossentropy')

t0 = time.time()
hist_den = denoiser.fit(
    make_denoise_dataset(x_tr_img, sigma=0.2),
    validation_data=make_denoise_dataset(x_val_img, sigma=0.2),
    epochs=20, verbose=2
)
den_train_time = time.time() - t0

# ---- Plot 4: Clean vs Noisy vs Denoised (sigma=0.2) ----
noisy_test = add_gaussian(x_test_img, 0.2)
denoised_test = denoiser.predict(noisy_test, batch_size=256, verbose=0)
show_grid([x_test, noisy_test, denoised_test],
          ["Clean", "Noisy (σ=0.2)", "Denoised"],
          n=10, figsize=(18, 3.5),
          suptitle="Plot 4 — Clean vs Noisy vs Denoised")

# ---- Plot 5: Noise level vs MSE / MAE / SSIM ----
noise_levels = [0.1, 0.2, 0.3]
rows = []
for s in noise_levels:
    nt = add_gaussian(x_test_img, s)
    dt = denoiser.predict(nt, batch_size=256, verbose=0)
    m, a, ss = compute_metrics(x_test, dt)
    rows.append((s, m, a, ss))
    
noise_df = pd.DataFrame(rows, columns=['Noise σ', 'MSE', 'MAE', 'SSIM']).set_index('Noise σ')
print("\n[Denoising] Noise vs Quality:\n", noise_df.round(4))

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
for i, (col, title) in enumerate(zip(['MSE', 'MAE', 'SSIM'],
                                     ['MSE (lower better)', 'MAE (lower better)', 'SSIM (higher better)'])):
    ax[i].plot(noise_df.index, noise_df[col], marker='o', color='C'+str(i))
    ax[i].set_xlabel('Gaussian noise σ'); ax[i].set_ylabel(col)
    ax[i].set_title(title); ax[i].grid(alpha=0.3)
plt.suptitle("Plot 5 — Noise Level vs Reconstruction Quality", fontsize=12)
plt.tight_layout(); plt.show()

# ---- Salt & pepper comparison (bonus, PDF §13) ----
print("\n[Salt & Pepper vs Gaussian — same denoiser]")
sp_rows = []
for p in [0.05, 0.10, 0.20]:
    nsp = add_salt_pepper(x_test_img, p)
    dsp = denoiser.predict(nsp, batch_size=256, verbose=0)
    m, a, ss = compute_metrics(x_test, dsp)
    sp_rows.append((p, m, a, ss))
sp_df = pd.DataFrame(sp_rows, columns=['p', 'MSE', 'MAE', 'SSIM']).set_index('p')
print(sp_df.round(4))

# =============================================================
# 4. VARIATIONAL AUTOENCODER (latent dim = 2)
# =============================================================
class Sampling(layers.Layer):
    def call(self, inputs):
        mu, log_var = inputs
        eps = tf.random.normal(shape=tf.shape(mu))
        return mu + tf.exp(0.5 * log_var) * eps

def build_vae(latent_dim=2):
    # Encoder
    enc_in = layers.Input(shape=(28, 28, 1))
    x = layers.Conv2D(32, 3, strides=2, activation='relu', padding='same')(enc_in)
    x = layers.Conv2D(64, 3, strides=2, activation='relu', padding='same')(x)
    x = layers.Flatten()(x)
    x = layers.Dense(128, activation='relu')(x)
    mu      = layers.Dense(latent_dim, name='mu')(x)
    log_var = layers.Dense(latent_dim, name='log_var')(x)
    z = Sampling()([mu, log_var])
    encoder = Model(enc_in, [mu, log_var, z], name='vae_encoder')

    # Decoder
    dec_in = layers.Input(shape=(latent_dim,))
    x = layers.Dense(7 * 7 * 64, activation='relu')(dec_in)
    x = layers.Reshape((7, 7, 64))(x)
    x = layers.Conv2DTranspose(64, 3, strides=2, activation='relu', padding='same')(x)
    x = layers.Conv2DTranspose(32, 3, strides=2, activation='relu', padding='same')(x)
    out = layers.Conv2DTranspose(1, 3, activation='sigmoid', padding='same')(x)
    decoder = Model(dec_in, out, name='vae_decoder')
    return encoder, decoder

class VAE(Model):
    def __init__(self, encoder, decoder, **kw):
        super().__init__(**kw)
        self.encoder = encoder
        self.decoder = decoder
        self.total_tr = tf.keras.metrics.Mean(name='total')
        self.rec_tr   = tf.keras.metrics.Mean(name='rec')
        self.kl_tr    = tf.keras.metrics.Mean(name='kl')
        self.total_vl = tf.keras.metrics.Mean(name='v_total')
        self.rec_vl   = tf.keras.metrics.Mean(name='v_rec')
        self.kl_vl    = tf.keras.metrics.Mean(name='v_kl')

    def _loss(self, x):
        mu, log_var, z = self.encoder(x, training=True)
        recon = self.decoder(z, training=True)
        rec = tf.reduce_mean(tf.reduce_sum(
            tf.keras.losses.binary_crossentropy(x, recon), axis=[1, 2]))
        kl = -0.5 * tf.reduce_mean(tf.reduce_sum(
            1 + log_var - tf.square(mu) - tf.exp(log_var), axis=-1))
        return rec + kl, rec, kl

    def train_step(self, data):
        # Keras passes (x, y) when you call fit(x, y). We only need x
        # (this is an autoencoder — target == input), so unwrap.
        x = data[0] if isinstance(data, (tuple, list)) else data
        with tf.GradientTape() as tape:
            total, rec, kl = self._loss(x)
        grads = tape.gradient(total, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))
        self.total_tr.update_state(total)
        self.rec_tr.update_state(rec)
        self.kl_tr.update_state(kl)
        return {'loss': self.total_tr.result(),
                'rec': self.rec_tr.result(),
                'kl': self.kl_tr.result()}

    def test_step(self, data):
        x = data[0] if isinstance(data, (tuple, list)) else data
        mu, log_var, z = self.encoder(x, training=False)
        recon = self.decoder(z, training=False)
        rec = tf.reduce_mean(tf.reduce_sum(
            tf.keras.losses.binary_crossentropy(x, recon), axis=[1, 2]))
        kl = -0.5 * tf.reduce_mean(tf.reduce_sum(
            1 + log_var - tf.square(mu) - tf.exp(log_var), axis=-1))
        total = rec + kl
        self.total_vl.update_state(total)
        self.rec_vl.update_state(rec)
        self.kl_vl.update_state(kl)
        return {'loss': self.total_vl.result(),
                'rec': self.rec_vl.result(),
                'kl': self.kl_vl.result()}
    @property
    def metrics(self):
        return [self.total_tr, self.rec_tr, self.kl_tr,
                self.total_vl, self.rec_vl, self.kl_vl]

enc, dec = build_vae(2)
vae = VAE(enc, dec)
vae.compile(optimizer=Adam(1e-3))

t0 = time.time()
hist_vae = vae.fit(
    x_tr_img, x_tr_img,
    validation_data=(x_val_img, x_val_img),
    epochs=20, batch_size=128, verbose=2
)
vae_train_time = time.time() - t0

# VAE reconstruction
mu_test, logvar_test, _ = enc.predict(x_test_img, batch_size=256, verbose=0)
vae_recon = dec.predict(mu_test, batch_size=256, verbose=0)
vae_mse, vae_mae, vae_ssim = compute_metrics(x_test, vae_recon)
print(f"[VAE] Test MSE={vae_mse:.5f}  MAE={vae_mae:.5f}  SSIM={vae_ssim:.4f}  "
      f"Params={vae.encoder.count_params() + vae.decoder.count_params():,}  "
      f"Time={vae_train_time:.1f}s")

# ---- Plot 9: VAE train/val loss ----
plt.figure(figsize=(8, 4))
plt.plot(hist_vae.history['loss'], label='Train Total')
plt.plot(hist_vae.history['val_loss'], label='Val Total')
plt.plot(hist_vae.history['rec'], '--', label='Train Rec')
plt.plot(hist_vae.history['kl'], ':', label='Train KL')
plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Plot 9 — VAE Training / Validation Loss')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

# ---- Plot 6: VAE latent space (2D) ----
plt.figure(figsize=(7, 6))
sc = plt.scatter(mu_test[:, 0], mu_test[:, 1], c=y_test, cmap='tab10', s=8, alpha=0.7)
plt.colorbar(sc, ticks=range(10), label='Digit')
plt.xlabel('z₁'); plt.ylabel('z₂')
plt.title('Plot 6 — VAE 2D Latent Space (colored by digit label)')
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

# ---- Plot 7: Random generated samples ----
rng = np.random.default_rng(SEED)
zs = rng.normal(size=(25, 2)).astype(np.float32)
generated = dec.predict(zs, verbose=0)
plt.figure(figsize=(7, 7))
for i in range(25):
    plt.subplot(5, 5, i + 1)
    plt.imshow(generated[i].squeeze(), cmap='gray', vmin=0, vmax=1)
    plt.axis('off')
plt.suptitle('Plot 7 — Randomly Generated Images (z ~ N(0, I))', fontsize=12)
plt.tight_layout(); plt.show()

# ---- Plot 8: Latent interpolation ----
# Pick two samples from different digit classes to make interpolation visually clear
idxA = np.where(y_test == 1)[0][0]
idxB = np.where(y_test == 7)[0][0]
zA, zB = mu_test[idxA], mu_test[idxB]
alphas = np.linspace(0, 1, 11).astype(np.float32)[:, None]
z_interp = (1 - alphas) * zA + alphas * zB
interp_imgs = dec.predict(z_interp, verbose=0)

plt.figure(figsize=(15, 2))
for i in range(11):
    plt.subplot(1, 11, i + 1)
    plt.imshow(interp_imgs[i].squeeze(), cmap='gray', vmin=0, vmax=1)
    plt.title(f'α={alphas[i,0]:.1f}', fontsize=8)
    plt.axis('off')
plt.suptitle(f'Plot 8 — Latent Interpolation (digit {y_test[idxA]} → digit {y_test[idxB]})', fontsize=12)
plt.tight_layout(); plt.show()

# =============================================================
# 5. RECONSTRUCTION-ERROR DISTRIBUTION + TOP-5 HIGH-ERROR SAMPLES
# =============================================================
errs = per_image_mse(x_test, cae_recon)     # use best model (CAE)

plt.figure(figsize=(8, 4))
plt.hist(errs, bins=40, color='steelblue', edgecolor='black')
plt.xlabel('Per-image MSE'); plt.ylabel('Count')
plt.title('Plot 10 — Test-Set Reconstruction-Error Distribution (CAE)')
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

top5 = np.argsort(errs)[-5:][::-1]
plt.figure(figsize=(12, 4))
for i, idx in enumerate(top5):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[idx].reshape(28, 28), cmap='gray'); plt.axis('off')
    plt.title(f'#{i+1} (MSE={errs[idx]:.4f})', fontsize=8)
    plt.subplot(2, 5, i + 6)
    plt.imshow(cae_recon[idx].reshape(28, 28), cmap='gray', vmin=0, vmax=1)
    plt.axis('off')
plt.suptitle('High-Error Samples — Original (top) vs Reconstruction (bottom)', fontsize=12)
plt.tight_layout(); plt.show()

# =============================================================
# 6. LATENT-DIMENSION STUDY  (PDF §27: 2, 8, 16, 32)
# =============================================================
print("\n[Latent-dim study — FC-AE]")
latent_results = []
for dz in [2, 8, 16, 32]:
    m = build_fc_ae(dz)
    m.compile(optimizer=Adam(1e-3), loss='binary_crossentropy')
    m.fit(x_tr_flat, x_tr_flat,
          validation_data=(x_val_flat, x_val_flat),
          epochs=10, batch_size=128, verbose=0)
    rec = m.predict(x_test_flat, batch_size=256, verbose=0)
    mm, aa, ss = compute_metrics(x_test, rec)
    latent_results.append((dz, mm, aa, ss))
    print(f"  z={dz:>2} | MSE={mm:.5f}  MAE={aa:.5f}  SSIM={ss:.4f}")

lat_df = pd.DataFrame(latent_results, columns=['Latent dim', 'MSE', 'MAE', 'SSIM']).set_index('Latent dim')
print("\nLatent-dim study table:\n", lat_df.round(4))

# =============================================================
# 7. CONSOLIDATED RESULTS TABLES
# =============================================================
summary = pd.DataFrame({
    'Model':        ['FC-AE (z=16)', 'CAE', 'Denoising CAE', 'VAE (z=2)'],
    'MSE':          [fc_mse, cae_mse, None, vae_mse],
    'MAE':          [fc_mae, cae_mae, None, vae_mae],
    'SSIM':         [fc_ssim, cae_ssim, None, vae_ssim],
    'Parameters':   [fc_ae.count_params(), cae.count_params(),
                     denoiser.count_params(),
                     vae.encoder.count_params() + vae.decoder.count_params()],
    'Train time(s)':[round(fc_train_time,1), round(cae_train_time,1),
                     round(den_train_time,1), round(vae_train_time,1)],
})
print("\n=== Consolidated Results ===")
print(summary.to_string(index=False))

print("\n=== VAE Losses (final epoch) ===")
print(f"Reconstruction loss: {hist_vae.history['rec'][-1]:.4f}")
print(f"KL loss            : {hist_vae.history['kl'][-1]:.4f}")
print(f"Total loss         : {hist_vae.history['loss'][-1]:.4f}")
print(f"Test MSE={vae_mse:.5f}  MAE={vae_mae:.5f}  SSIM={vae_ssim:.4f}")
print("\nAll done.")
```
